```python
import pandas as pd
import matplotlib.pyplot as plt

# ==================================================
# LEITURA DA BASE ENEM
# ==================================================

enem = pd.read_excel('/content/Enem_2024_Amostra_Perfeita.xlsx')

# ==================================================
# QUESTÃO 1
# GRÁFICO DE BARRAS - TP_COR_RACA
# ==================================================

freq_raca = (enem['TP_COR_RACA'].value_counts().sort_index())

plt.figure(figsize=(8,5))
freq_raca.plot(kind='bar')

plt.title('Distribuição da variável TP_COR_RACA')
plt.xlabel('TP_COR_RACA')
plt.ylabel('Frequência')
plt.grid(axis='y', alpha=0.3)

plt.show()

# ==================================================
# QUESTÃO 2
# NOTA_MT_MATEMATICA > 680
# ==================================================

alunos_680 = enem[enem['NOTA_MT_MATEMATICA'] > 680]

# Tabela de frequência absoluta
freq_abs = alunos_680['SG_UF_PROVA'].value_counts()

# Tabela final
tabela_freq = pd.DataFrame({'Frequencia_Absoluta': freq_abs,'Frequencia_Relativa (%)':round(freq_abs/freq_abs.sum()*100, 2)})

tabela_freq = tabela_freq.reset_index()
tabela_freq.columns = ['SG_UF_PROVA','Frequencia_Absoluta','Frequencia_Relativa (%)']

print("\nTabela de Distribuição de Frequência")
display(tabela_freq)
```
