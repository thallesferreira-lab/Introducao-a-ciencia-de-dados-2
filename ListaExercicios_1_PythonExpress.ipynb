```python
# 2-Elabore um código que peça um número e então mostre a mensagem: "O número informado foi [número]."
numero=input("digite um numero")
print(f"o numero informado foi {numero}")
```

```python
# 3-Elabore um código que peça dois números e imprima a soma.
numero1=int(input("digite um numero"))
numero2=int(input("digite um numero"))
print(f"a soma dos numeros é {numero1+numero2}")
```

```python
# 4-Elabore um código que peça as 4 notas bimestrais e mostre a média
nota1=float(input("qual sua nota?"))
nota2=float(input("qual sua nota?"))
nota3=float(input("qual sua nota?"))
nota4=float(input("qual sua nota?"))
media=(nota1+nota2+nota3+nota4)/4
print(f"sua media é {media}")
```

```python
# 5-Elabore um código que converta metros para centímetros.

metros=float(input("digite o valor em metros"))
centimetros=metros*100
print(f"o valor em centimetros é {centimetros}")
```

```python
# 6- Elabore um código que peça o raio de um círculo, calcule e mostre sua área.
raio=float(input("digite o raio do circulo"))
area=3.14*(raio**2)
print(f"a area do circulo é {area}")
```

```python
# 7-Elabore um código que calcule a área de um quadrado, em seguida mostre o dobro desta área para o usuário.

lado = float(input("Digite o comprimento do lado do quadrado: "))
area = lado ** 2
dobro_da_area = area * 2
print(f"A área do quadrado é {area} e o dobro dessa área é {dobro_da_area}.")
```

```python
# 8-Elabore um código que pergunte quanto você ganha por hora e o número de horas trabalhadas no mês. Calcule e mostre o total do seu salário no referido mês em uma mensagem amigável.
salario_hora=float(input("quanto voce ganha por hora?"))
horas_trabalhadas=float(input('quantas horas voce trabalha no mes? ;-;'))
salario_mes=salario_hora*horas_trabalhadas
print(f"seu salario no mes é {salario_mes}!!!!!!!!!!")
```

```python
# 9-Elabore um código que peça a temperatura em graus Fahrenheit, transforme e mostre a temperatura em graus Celsius. ** Onde C = 5 * ((F-32) / 9)**.
temperatura_f=float(input("digite a temperatura em fahrenheit"))
graus_celsius= 5*((temperatura_f-32)/9)
print(f'a temperatura em graus celsius é {graus_celsius}')
```

```python
# 10-Elabore um código que peça a temperatura em graus Celsius, transforme e mostre em graus Fahrenheit
graus_celsius = float(input("digite a temperatura em graus celsius"))
temperatura_f = (graus_celsius * 9/5) + 32
print(f"temperatura em graus Fahrenheit é {temperatura_f}")
```

```python
# 11-Elabore um código que peça 2 números inteiros e um número real. Calcule e mostre:
# A) o produto do dobro do primeiro com metade do segundo .
# B) a soma do triplo do primeiro com o terceiro.
# C) o terceiro elevado ao cubo.

numero1 = float(input("apresente um numero inteiro"))
numero2 = float(input("apresente um numero inteiro"))
numero3 = float(input("apresente um numero real"))
A = (numero1 * 2) * (numero2/2)
B = (numero1 * 3) + numero3
C = numero3 ** 3
print(f"o o produto do dobro do primeiro com metade do segundo é {A}")
print(f"a soma do triplo do primeiro com o terceiro é {B}")
print(f"o terceiro elevado ao cubo {C}")
```

```python
# 12- Elabore um código que receba os dados de entrada a altura de uma pessoa, e informe qual seria o peso ideal, usando a seguinte fórmula: (72.7*altura) - 58
altura = float(input("digite sua altura"))
peso_ideal = (72.7  * altura) - 58
print(f"seu peso ideal é {peso_ideal}")
```
