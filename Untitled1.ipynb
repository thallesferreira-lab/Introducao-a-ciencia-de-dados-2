```python
# 10. Número par ou ímpar
num = int(input("Digite um número inteiro: "))
if num % 2 == 0:
    print("PAR")
else:
    print("ÍMPAR")
```

```python
# 11. Classificação por idade
idade = int(input("Digite sua idade: "))
if idade < 18:
    print("Menor de idade")
elif idade < 60:
    print("Adulto")
else:
    print("Idoso")
```

```python
# 12. Condicional com x e y
x = 5
y = 8
if x > 0 and y < 10:
    print("Valores válidos")
else:
    print("Valores fora do limite")
```

```python
# 13. Múltiplos de 3 de 0 a 30
for i in range(0, 31):
    if i % 3 == 0:
        print(i)
```

```python
# 14. Lista de disciplinas
disciplinas = ["Matemática", "Estatística", "Programação"]
for d in disciplinas:
    print(f"Eu preciso estudar {d}")
```

```python
# 15. Contagem regressiva
contador = 10
while contador >= 0:
    print(contador)
    contador -= 1
print("Decolar!")
```

```python
# 16. List comprehension (cubos dos pares)
cubos = [x**3 for x in range(11) if x % 2 == 0]
print(cubos)
```

```python
# 17. Função sem retorno
def saudacao_personalizada(nome):
    print(f"Seja muito bem-vindo(a) ao mundo dos dados, {nome}!")
saudacao_personalizada("Thalles")
```

```python
# 18. Função com desconto
def calcula_desconto(valor, desconto):
    return valor - (valor * desconto / 100)

print(calcula_desconto(100, 10))
```

```python

# 19. Classificação de nota
def classifica_nota(nota):
    if nota >= 7:
        return "Aprovado"
    elif nota >= 5:
        return "Recuperação"
    else:
        return "Reprovado"

print(classifica_nota(6.5))
```

```python

# 20. Contar positivos, negativos e zeros
lista = [10, -5, 0, 3, -1, 0, 7]
positivos = 0
negativos = 0
zeros = 0

for n in lista:
    if n > 0:
        positivos += 1
    elif n < 0:
        negativos += 1
    else:
        zeros += 1
print(f"Positivos: {positivos}")
print(f"Negativos: {negativos}")
print(f"Zeros: {zeros}")
```
